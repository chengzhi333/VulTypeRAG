```python
def generate_latex_table(data, output_file='latex_table.txt'):
    # Define colors for different k values (supports up to 6 k values)
    color_map = {
        1: 'gray!20',
        3: 'gray!20',
        5: 'blue!20',
        10: 'green!20',
        15: 'yellow!20',
        20: 'orange!20',
        25: 'red!20'
    }
    col2_header = list(data[0].keys())[1]
    # Group data by k value
    grouped_data = {}
    for entry in data:
        k = entry['k']
        if k not in grouped_data:
            grouped_data[k] = []
        grouped_data[k].append(entry)
    
    # Generate LaTeX table header
    latex_code = f"""\\renewcommand{{\\arraystretch}}{{1.4}}

\\begin{{longtable}}[ht]{{|>{{\\columncolor{{gray!20}}}}c|c|c|c|c|c|}}
    \\hline
    \\rowcolor{{gray!40}} \\textbf{{k}} & \ textbf{{{col2_header.capitalize()}}} &  {{\\textbf{{Accuracy}}}} & \\textbf{{Precision}} & \\textbf{{Recall}} & \\textbf{{F1-score}} \\\\ \\hline \\endhead 
"""
    
    # Generate table rows for each k group
    for k, entries in grouped_data.items():
        color = color_map.get(k, 'gray!20')  # Default to gray if k not in color map
        
        # Generate rows for this k group
        for i, entry in enumerate(entries):
            if i == len(entries) - 1:
                # Last row in group - use multirow
                latex_code += f"""    \\rowcolor{{{color}}}
    \\multicolumn{{1}}{{|>{{\\columncolor{{{color}}}}}c|}}{{\\multirow{{-{len(entries)}}}{{*}}{{{k}}}}} & {entry[col2_header]}  & {entry['accuracy']:.4f} & {entry['precision']:.4f} & {entry['recall']:.4f} & {entry['f1']:.4f} \\\\ \\hline
"""
            else:
                # Regular row in group
                latex_code += f"""    \\rowcolor{{{color}}}
    \\cellcolor{{{color}}}\\vphantom{{\\multirow{{-{len(entries)}}}{{*}}{{{k}}}}} & {entry[col2_header]}  & {entry['accuracy']:.4f} & {entry['precision']:.4f} & {entry['recall']:.4f} & {entry['f1']:.4f} \\\\ \\cline{{2-6}}
"""     
        latex_code += '\n'
    
    # Add table footer
    latex_code += f"""    \\caption{{Performance metrics by $k$ and {col2_header}}}
\\end{{longtable}}
"""
    
    # Save to file
    with open(output_file, 'w') as f:
        f.write(latex_code)
    
    return latex_code
```

```python
import os
import matplotlib.pyplot as plt
import pandas as pd
def table_to_svg(pretty_table, svg_filename):
    df = pd.DataFrame(pretty_table._rows, columns=pretty_table.field_names)

    # Estimate text width for each column
    col_widths = []
    font_size = 12
    char_width = font_size * 1.4  # average width per character

    for col in df.columns:
        max_len = max([len(str(val)) for val in df[col]] + [len(col)])
        col_widths.append(max_len * char_width)

    total_width = sum(col_widths)
    col_rel_widths = [w / total_width for w in col_widths]

    # Set up figure width/height
    fig_height = 1 + 0.7 * len(df)  # Increased for taller rows
    fig_width = total_width / 100
    fig, ax = plt.subplots(figsize=(fig_width, fig_height))
    ax.axis('off')

    # Create table
    mpl_table = ax.table(
        cellText=df.values,
        colLabels=df.columns,
        cellLoc='center',
        loc='center'
    )

    mpl_table.auto_set_font_size(False)
    mpl_table.set_fontsize(font_size)
    mpl_table.scale(1, 2.5)  # Increased vertical scaling for taller rows

    # Style cells
    for (row, col), cell in mpl_table.get_celld().items():
        cell.set_linewidth(0)
        if row == 0:
            cell.set_text_props(weight='bold', color='white')
            cell.set_facecolor('#4a4a4a')
        else:
            cell.set_facecolor('#d5d5d5' if row % 2 == 0 else '#ebebeb')
        cell.set_width(col_rel_widths[col])

    fig.savefig(
            os.path.join('tmp', svg_filename),
            format='svg',
            bbox_inches='tight',  # Removes extra whitespace
            pad_inches=0,         # Removes padding
            transparent=True      # Optional: removes background
        )
    plt.close(fig)
```

```python
import statistics as stat
from prettytable import PrettyTable

def gen_table(data):
    table = PrettyTable(
        data[0].keys()
    )
    table.title = "True Positive Sample Statistics" 
    table.add_rows(
        [[str(val) for val in dat.values()] for dat in data]
    )
    return table
```

```python
baseline_data = [
    {'CWE View': '1000', 'Accuracy': 0.4628, 'Precision': 0.6278, 'Recall': 0.4628, 'F1-Score': 0.4885},
    {'CWE View': '1003', 'Accuracy': 0.4358, 'Precision': 0.5951, 'Recall': 0.4358, 'F1-Score': 0.4628},
]

table = gen_table(baseline_data)
# Generate and save the LaTeX table
table_to_svg(table, 'baseline_metrics_table.svg')
print("LaTeX table has been generated and saved to 'tmp/baseline_metrics_table.svg'")
```
